# FairRoute scaling runner

Builds and validates a resumable manifest. Set `RUN_SIM=1` only after filling the project-specific simulator command.

```python
from pathlib import Path
import csv, hashlib, json, os
ROOT=Path.cwd(); TRACE_ROOT=ROOT/'workloads/fairroute/scaling'; OUT=ROOT/'outputs/fairroute/scaling'; OUT.mkdir(parents=True, exist_ok=True)
POLICIES=('RR','LOAD','PRISM','D2LPM','FAIRROUTE')
ALL_CELLS=((4,.5),(8,.8),(16,1.0),(32,1.25),(64,1.5),(4,1.0),(8,1.0),(32,1.0),(64,1.0),(16,.5),(16,.8),(16,1.25),(16,1.5))
MAIN_MATRIX_CELLS={(16,.5),(16,.8),(16,1.0),(16,1.25),(16,1.5)}
CELLS=tuple(cell for cell in ALL_CELLS if cell not in MAIN_MATRIX_CELLS)
rows=[]
for workload in ('w1','w3'):
  for k,rho in CELLS:
    for policy in POLICIES:
      for seed in (100,):
        rows.append({'run_id':f'{workload}_K{k}_L{rho}_s{seed}_{policy.lower()}','workload':workload,'K':k,'rho':rho,'seed':seed,'policy':policy,'status':'pending'})
assert len(rows)==80, len(rows)
with (OUT/'runs_manifest.csv').open('w',newline='') as handle:
  writer=csv.DictWriter(handle,fieldnames=rows[0]); writer.writeheader(); writer.writerows(rows)
print(f'manifest rows: {len(rows)}; execution is intentionally opt-in')
```
